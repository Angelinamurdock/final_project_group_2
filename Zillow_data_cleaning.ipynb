```python
# Import Modules
import numpy as np
import pandas as pd
```

**Import and clean zillow CSV and calculate average annual home price by neighborhood**

```python
# Read in and View CSV
zillow_df = pd.read_csv("Resources/Neighborhood_zillow_real_estate_data.csv")
zillow_df.head()
```

```python
# Pull only the data for Denver
denver_real_estate_neighborhood_df = zillow_df.query("City == 'Denver'")
denver_real_estate_neighborhood_df.head()
```

```python
# Drop Columns not needed
denver_real_estate_neighborhood_df = denver_real_estate_neighborhood_df.drop(columns=["SizeRank", "RegionType", "StateName", "State", "Metro", "CountyName"], axis=1)
denver_real_estate_neighborhood_df.head()
```

```python
#Rename Region Name Column
denver_real_estate_neighborhood_df.rename(columns={"RegionName":"Neighborhood"}, inplace=True)
denver_real_estate_neighborhood_df.head()
                                          
```

```python
# # Drop Null Values
# denver_real_estate_neighborhood_df = denver_real_estate_neighborhood_df.dropna
```

```python
#Check Data Types
denver_real_estate_neighborhood_df.dtypes
```

```python
# Step 1: Melt all date columns (all columns after the first 3)
zillow_melted_df = denver_real_estate_neighborhood_df.melt(
    id_vars=["RegionID", "Neighborhood", "City"],
    var_name="Date",
    value_name="Value"
)

# Step 2: Convert 'Date' to datetime
zillow_melted_df["Date"] = pd.to_datetime(zillow_melted_df["Date"], errors='coerce')

# Step 3: Drop rows where 'Date' couldn't be parsed
zillow_melted_df = zillow_melted_df.dropna(subset=["Date"])

# Step 4: Extract year
zillow_melted_df["Year"] = zillow_melted_df["Date"].dt.year

# Step 5: Group by RegionID, Neighborhood, City, and Year
zillow_grouped_df = zillow_melted_df.groupby(
    ["RegionID", "Neighborhood", "City", "Year"]
)["Value"].mean().round(0).reset_index()

zillow_grouped_df
```

```python
#Rename Value(Average) Column
zillow_grouped_df.rename(columns={"Value": "Avg Annual Home Value"},inplace=True)
zillow_grouped_df.head()
```

```python
#Export CSV
zillow_grouped_df.to_csv("Resources/zillow_yearly.csv")
```

**Calculate average quarterly home price by neighborhood**

```python
# Step 1: Melt the data
zillow_quarterly_melted_df = denver_real_estate_neighborhood_df.melt(
    id_vars=["RegionID", "Neighborhood", "City"],
    var_name="Date",
    value_name="Value"
)

# Step 2: Convert 'Date' to datetime
zillow_quarterly_melted_df["Date"] = pd.to_datetime(zillow_quarterly_melted_df["Date"], errors='coerce')

# Step 3: Drop invalid dates
zillow_quarterly_melted_df = zillow_quarterly_melted_df.dropna(subset=["Date"])

# Step 4: Extract Year and Quarter
zillow_quarterly_melted_df["Year"] = zillow_quarterly_melted_df["Date"].dt.year
zillow_quarterly_melted_df["Quarter"] = zillow_quarterly_melted_df["Date"].dt.quarter

# Step 5: Group by year and quarter and get average value
grouped_quarterly_df = zillow_quarterly_melted_df.groupby(
    ["RegionID", "Neighborhood", "City", "Year", "Quarter"]
)["Value"].mean().round(2).reset_index()

grouped_quarterly_df
```

```python
grouped_quarterly_df.rename(columns={"Value": "Avg Annual Home Value"},inplace=True)
grouped_quarterly_df.head()
```

```python
# Export to CSV
grouped_quarterly_df.to_csv("Resources/zillow_quarterly.csv")
```
