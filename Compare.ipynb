```python
import matplotlib.pyplot as plt
import numpy as np
import re
```

```python
# Data
models = ['Zillow Regression', 'Prophet', 'Keras', 'Keras Quarterly']
rmse_values = [82034.22, 71127.43, 83013.98, 39307.90]
r2_values = [0.8659, 0.9265, 0.8809, 0.9692]

x = np.arange(len(models))
width = 0.35

# Create bar chart
fig, ax1 = plt.subplots(figsize=(12, 6))

# RMSE bars
bar1 = ax1.bar(x - width/2, rmse_values, width, label='RMSE', color='skyblue')
ax1.set_ylabel('RMSE (lower is better)')
ax1.set_xlabel('Model')
ax1.set_title('Model Performance Comparison: RMSE and R²')
ax1.set_xticks(x)
ax1.set_xticklabels(models, rotation=15)
ax1.tick_params(axis='y')

# Twin axis for R²
ax2 = ax1.twinx()
bar2 = ax2.bar(x + width/2, r2_values, width, label='R² Score', color='gray')
ax2.set_ylabel('R² Score (higher is better)')
ax2.set_ylim(0, 1.1)

# Add value labels on top of bars
for bars in [bar1, bar2]:
    for bar in bars:
        height = bar.get_height()
        ax = ax1 if bars is bar1 else ax2
        ax.text(bar.get_x() + bar.get_width()/2., height * 1.01,
                f'{height:,.2f}', ha='center', va='bottom', fontsize=9)

# Legend
fig.legend(loc='upper center', bbox_to_anchor=(0.5, 1.08), ncol=2)
plt.tight_layout()
plt.show()


```
