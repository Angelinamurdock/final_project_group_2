```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter


```

```python
# Load your CSV (update the path if needed)
df = pd.read_csv("Resources/zillow_quarterly.csv")


```

```python
# Filter for unique Denver neighborhoods
denver_neighborhoods = df[df["City"].str.lower() == "denver"][["Neighborhood"]].drop_duplicates()

```

```python
# Set up geocoder
geolocator = Nominatim(user_agent="denver_neighborhood_locator")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

```

```python
# Define a function to get coordinates
def get_coords(neighborhood):
    try:
        location = geocode(f"{neighborhood}, Denver, CO")
        return pd.Series([location.latitude, location.longitude]) if location else pd.Series([None, None])
    except:
        return pd.Series([None, None])

```

```python
# Apply to the dataset
denver_neighborhoods[["Latitude", "Longitude"]] = denver_neighborhoods["Neighborhood"].apply(get_coords)

# Drop missing values
denver_neighborhoods = denver_neighborhoods.dropna()

# Save result
denver_neighborhoods.to_csv("Resources/denver_neighborhood_coordinates.csv", index=False)
print("Coordinates saved to denver_neighborhood_coordinates.csv")
```
